# Preprocessing Data

## 1. Pengertian Preprocessing
Preprocessing adalah tahap awal dalam pengolahan data sebelum masuk ke proses analisis atau machine learning.

Tujuannya:

Membersihkan data (data cleaning)
Menangani missing value
Menyamakan skala data (normalisasi)
Mengubah format data agar siap diproses

Oleh karena itu, preprocessing dilakukan agar kualitas data meningkat sehingga hasil analisis atau model yang dihasilkan menjadi lebih akurat dan juga valid.

## 2. Tujuan Data Preprocessing
Tujuan data preprocessing adalah untuk mempersiapkan data mentah agar menjadi lebih bersih, rapi, dan siap digunakan dalam proses analisis atau pemodelan (misalnya machine learning).

- Mengatasi data kosong & error

- Menyeragamkan format

- Melakukan normalisasi

- Meningkatkan akurasi hasil analisis/model


## 3. Tahapan dalam Data Preprocessing

### 3.1 Data Cleaning
mengatasi missing value, noise, dan data duplikat

Contoh masalah yang diperbaiki:
- Missing value
- Data duplikat
- Kesalahan penulisan
- Outlier

- Data: umur = kosong → diisi rata-rata (misal 20)
- Data duplikat dihapus
- Nilai salah (umur = 200) diperbaiki/dihapus

Nilai umur yang kosong dapat diisi dengan rata-rata umur.



### 3.2 Data Integration
mengatasi missing value, noise, dan data duplikat

- Menggabungkan data mahasiswa dari file A (nilai) dan file B (absensi) menjadi satu dataset

Semua data tersebut digabungkan untuk analisis yang lebih lengkap.


### 3.3 Data Transformation
normalisasi, encoding, dan perubahan format

- Normalisasi nilai (0–100 → 0–1)
Mengubah kategori:
umur  = 18, setelah normalisasi = 0

Contoh normalisasi:

| Umur | Setelah Normalisasi |
|-----|---------------------|
| 18 | 0.2 |
| 25 | 0.9 |

Tujuannya agar nilai memiliki skala yang sama.

### 3.4 Data Reduction
mengurangi fitur yang tidak penting

- Dari banyak fitur (nama, NIM, alamat, umur, nilai) → ambil yang penting saja (umur, nilai)
Contoh:
Dataset dengan 100 atribut dapat dipilih menjadi hanya 10 atribut yang paling relevan.


### 3.5 Data Discretization
mengubah data kontinu menjadi kategori

Umur:
- 17–20 = Remaja
- 21–25 = Dewasa

setiap tahap ada contoh nyata bagaimana data diubah supaya lebih siap dipakai.
Hal ini sering digunakan pada algoritma klasifikasi tertentu.


## 4. Contoh Alur Preprocessing

- Tahap pertama adalah data cleaning, yaitu membersihkan data dengan cara mengisi nilai yang kosong, memperbaiki data yang salah, dan menghapus duplikasi.

- Selanjutnya data transformation, di mana data diubah agar lebih mudah diproses, misalnya dengan normalisasi atau mengubah data kategorikal menjadi angka.

- Setelah itu dilakukan data reduction, yaitu menyederhanakan data dengan memilih fitur yang penting saja agar proses lebih efisien.

- Terakhir, data menjadi siap digunakan, artinya sudah bersih, rapi, dan bisa langsung dipakai untuk analisis atau pemodelan.

Setelah proses ini selesai, data siap digunakan untuk proses **data mining** seperti:
- Klasifikasi
- Clustering
- Prediksi
- Association rule

## 5. Kesimpulan
Data preprocessing adalah proses penting untuk mengubah data mentah menjadi data yang bersih, terstruktur, dan siap digunakan, melalui tahap pembersihan, transformasi, dan penyederhanaan, sehingga hasil analisis atau model menjadi lebih akurat dan optimal.

# Normalisasi Data

## Pengertian Normalisasi Data
Normalisasi data adalah proses mengubah nilai data ke dalam skala tertentu (biasanya lebih kecil atau seragam) tanpa mengubah informasi aslinya, agar data lebih mudah dibandingkan dan diproses oleh algoritma.

- KNN (K-Nearest Neighbor)
- WKNN (Weighted KNN)
- Clustering
- Cosine Similarity

Jika data memiliki skala yang berbeda, misalnya **usia (18–23)** dan **IPK (2–5)**, maka atribut dengan nilai yang lebih besar dapat mendominasi perhitungan.


#Contoh

|---|---|---|

| Semester | 5 | 3 |
| IPK | 2.22 | 3.22 |
| Jumlah_Organisasi | 1 | 1 |
| Jenis_Kelamin | Perempuan | Perempuan |


Pada proses normalisasi biasanya **hanya atribut numerik yang dinormalisasi**, seperti:

- Usia  
- Semester
- IPK
- Jumlah Organisasi
- Jenis kelamin

# 1. Min-Max Normalization

Min-Max Normalization digunakan untuk **mengubah nilai data ke dalam rentang 0 sampai 1**.

## Rumus

$
X' = (X - Xmin) / (Xmax - Xmin)
$

### Contoh pada atribut Usia

- Xmin = 18
- Xmax = 25

| Mahasiswa | Perhitungan | Hasil |
|---|---|---|
| 1 | (18 − 18) / (25 − 18) | 0 |
| 2 | (25 − 18) / (25 − 18) | 1 |

### Contoh pada atribut IPK

- Xmin = 2.22
- Xmax = 3.72

| Mahasiswa | Perhitungan | Hasil |
|---|---|---|
| 1 | (2.22 − 2.22) / (3.72 − 2.22) | 0 |
| 2 | (3.72 − 2.22) / (3.72 − 2.22) | 1 |

### Hasil Normalisasi Min-Max

| Atribut | Mahasiswa 1 | Mahasiswa 2 |
|---|---|---|
| Usia | 0 | 1 |
| Semester | 1 | 0 |
| IPK | 0 | 1 |
| Jumlah_Organisasi | 0 | 1 |


# Min-Max Normalization dengan Rentang Baru (New Min dan New Max)

Min-Max Normalization (dengan rentang baru) adalah metode normalisasi yang mengubah nilai data ke dalam rentang tertentu yang diinginkan (misalnya 0–1 atau 0–100).

Rentang baru ini disebut:

- **NewMin** → nilai minimum baru
- **NewMax** → nilai maksimum baru

Metode ini berguna ketika suatu algoritma membutuhkan skala tertentu, misalnya:

- 0 – 10
- -1 – 1
- 1 – 100


## Rumus

$
X' = (X - Xmin) / (Xmax - Xmin) × (NewMax - NewMin) + NewMin
$

Keterangan:
- X = nilai awal
- Xmin & Xmax = acuan dari data lama
- NewMin & NewMax = tujuan skala baru
- X' = hasil akhir setelah disesuaikan ke skala baru

## Contoh

Data usia:

- 18
- 25

Nilai minimum = 18  
Nilai maksimum = 25  

Misalkan ingin mengubah data ke rentang **0 sampai 10**.

$
X' = (X - 18) / (25 - 18) × (10 - 0)
$

### Perhitungan

| Mahasiswa | Perhitungan | Hasil |
|---|---|---|
| 1 | (18 − 18) / (25 − 18) × 10 | 0 |
| 2 | (25 − 18) / (25 − 18) × 10 | 10 |


## Contoh pada atribut IPK
Data IPK:

- 2.22
- 3.72

Nilai minimum = 2.22  
Nilai maksimum = 3.72  

Rentang baru yang digunakan **0 – 10**

$
X' = (X - 2.22) / (3.72 - 2.22) × 10
$

### Perhitungan

| Mahasiswa | Perhitungan | Hasil |
|---|---|---|
| 1 | (2.22 − 2.22) / (3.72 − 2.22) × 10 | 0 |
| 2 | (3.72 − 2.22) / (3.72 − 2.22) × 10 | 10 |


Dengan demikian, data dapat disesuaikan dengan kebutuhan algoritma atau analisis yang digunakan tanpa mengubah hubungan antar nilai dalam dataset.



K-Nearest Neighbor (KNN)
1. Pengertian K-Nearest Neighbor (KNN)

K-Nearest Neighbor (KNN) adalah salah satu metode dalam machine learning yang digunakan untuk menyelesaikan masalah klasifikasi dan regresi. Cara kerja algoritma ini didasarkan pada asumsi bahwa suatu data baru akan memiliki karakteristik yang mirip dengan data di sekitarnya.
TiNgkat kemiripan antar data ditentukan dengan menghitung jarak antar titik data dalam ruang fitur (feature space). Data dengan jarak paling kecil dianggap sebagai tetangga terdekat.

Nilai K menunjukkan jumlah tetangga yang digunakan sebagai acuan dalam menentukan hasil prediksi.

---

2. Mekanisme Kerja KNN

Secara garis besar, proses KNN dilakukan melalui langkah berikut:
Menentukan jumlah tetangga (K) yang akan digunakan.
Menghitung jarak antara data uji dengan seluruh data latih.
Mengurutkan jarak dari nilai yang paling kecil.
Mengambil K data terdekat sebagai referensi.
Menentukan hasil:
Klasifikasi → berdasarkan mayoritas kelas tetangga
Regresi → berdasarkan nilai rata-rata tetangga

---
3. Euclidean Distance
Dalam KNN, pengukuran jarak yang umum digunakan adalah Euclidean Distance. Metode ini menghitung jarak lurus antara dua titik dalam ruang berdimensi banyak.

Rumus Euclidean Distance:

$
d(p,q) = \sqrt{\sum_{i=1}^{n} (p_i - q_i)^2}
$

Keterangan:
- \(d(p,q)\) = jarak antara data \(p\) dan \(q\)
- \(p_i\) = nilai atribut ke-i pada data \(p\)
- \(q_i\) = nilai atribut ke-i pada data \(q\)
- \(n\) = jumlah atribut

Semakin kecil nilai jarak yang dihasilkan, maka semakin tinggi tingkat kemiripan antara kedua data tersebut.
Semakin kecil hasil jarak, maka kedua data tersebut semakin mirip.

4. Missing Value pada Data Numerik
Dalam dataset, sering ditemukan nilai yang tidak tersedia atau disebut missing value. Hal ini bisa terjadi karena data tidak lengkap, kesalahan input, atau proses pengumpulan yang kurang sempurna.
Jika tidak ditangani, missing value dapat menurunkan kualitas analisis dan mempengaruhi hasil model.

Salah satu cara untuk mengatasinya adalah menggunakan metode KNN Imputation.

---

5. KNN Imputation
KNN Imputation adalah teknik untuk mengisi nilai yang hilang dengan memanfaatkan data lain yang memiliki kemiripan tinggi.
Pendekatan ini menggunakan konsep KNN, di mana nilai yang kosong akan diperkirakan berdasarkan nilai dari tetangga terdekat. Tingkat kedekatan dihitung menggunakan metode jarak, seperti Euclidean Distance.

---
6. Tahapan Mengisi Missing Value dengan KNN

Langkah-langkah yang dilakukan adalah:
- Menentukan data yang memiliki nilai kosong.
- Menghitung jarak antara data tersebut dengan data lain yang lengkap.
- Mengurutkan hasil jarak dari yang terkecil.
- Memilih K tetangga terdekat.
- Menghitung estimasi nilai (biasanya dengan rata-rata).
- Mengganti nilai yang hilang dengan hasil perhitungan tersebut.

---

7. Kelebihan dan Kekurangan KNN untuk Imputasi
Kelebihan:
- Mudah dipahami dan diimplementasikan
- Memanfaatkan data yang paling mirip sehingga hasil cukup akurat
- Bisa digunakan pada berbagai jenis data numerik

Kekurangan:
- Kurang efisien untuk data berukuran besar
- Sensitif terhadap perbedaan skala (perlu normalisasi)
- Hasil dipengaruhi oleh pemilihan nilai K

---

8. Kesimpulan:
KNN merupakan metode berbasis kemiripan data yang banyak digunakan dalam analisis data. Dengan bantuan perhitungan jarak seperti Euclidean Distance, algoritma ini dapat digunakan tidak hanya untuk klasifikasi dan regresi, tetapi juga untuk mengisi missing value melalui teknik KNN Imputation. Prinsip utamanya adalah memanfaatkan data yang paling dekat untuk memperkirakan nilai yang hilang.

# Weighted K-Nearest Neighbor(WKNN)

## Pengertian WKNN

Pengertian WKNN (Weighted K-Nearest Neighbor)

WKNN (Weighted K-Nearest Neighbor) adalah pengembangan dari algoritma KNN (K-Nearest Neighbor) yang digunakan untuk klasifikasi atau prediksi, dengan menambahkan bobot (weight) pada setiap tetangga berdasarkan jaraknya.

Penjelasan Sederhana
- KNN biasa → semua tetangga dianggap punya pengaruh sama

- WKNN → tetangga yang lebih dekat punya pengaruh lebih besar
Dengan pendekatan ini, WKNN mampu menghasilkan keputusan yang lebih akurat karena mempertimbangkan tingkat kedekatan antar data secara lebih detail.

---

## Konsep Dasar WKNN
WKNN (Weighted K-Nearest Neighbor) adalah metode prediksi yang menentukan hasil berdasarkan K data terdekat, dengan memberikan bobot lebih besar pada data yang jaraknya lebih dekat.

---

## Rumus Perhitungan pada WKNN

### 1. Rumus Euclidean Distance

Rumus ini digunakan untuk menghitung jarak antara data uji dan data latih.

$
d(x,y) = \sqrt{\sum_{i=1}^{n} (x_i - y_i)^2}
$

Keterangan:

- \( d(x,y) \) = jarak antara dua data  
- \( x_i \) = nilai atribut pada data uji  
- \( y_i \) = nilai atribut pada data latih  
- \( n \) = jumlah atribut yang digunakan  
WKNN (Weighted K-Nearest Neighbor) adalah metode prediksi yang menentukan hasil berdasarkan K data terdekat, dengan memberikan bobot lebih besar pada data yang jaraknya lebih dekat.
---

### 2. Rumus Bobot pada WKNN

Setelah jarak dihitung, setiap tetangga akan diberikan bobot berdasarkan jaraknya.

$
w_i = \frac{1}{d_i}
$

Keterangan:

- \( w_i \) = bobot tetangga ke-i  
- \( d_i \) = jarak tetangga ke-i terhadap data uji  

Semakin kecil nilai jarak, maka bobot yang dihasilkan akan semakin besar.

---

## Tahapan Proses WKNN

Secara umum, proses kerja algoritma WKNN terdiri dari beberapa tahap:

1. Menyiapkan dataset yang akan digunakan.
2. Melakukan preprocessing data seperti normalisasi agar skala atribut menjadi seragam.
3. Menghitung jarak antara data uji dan seluruh data latih menggunakan metode pengukuran jarak.
4. Mengurutkan jarak dari yang paling kecil.
5. Menentukan jumlah tetangga terdekat (nilai **K**).
6. Menghitung bobot setiap tetangga berdasarkan jaraknya.
7. Melakukan proses voting berbobot.
8. Menentukan kelas dengan total bobot terbesar sebagai hasil prediksi.

---

## Kelebihan WKNN
- Lebih akurat karena menggunakan bobot berdasarkan jarak
- Mengurangi pengaruh data yang jauh
- Lebih sensitif terhadap pola lokal
- Cocok untuk data yang tidak merata
- Dapat digunakan untuk klasifikasi dan regresi
---

## Kekurangan WKNN
- Perhitungan lebih kompleks karena ada pembobotan
- Sensitif terhadap jarak yang sangat kecil (bisa menghasilkan bobot sangat besar)
- Tetap bergantung pada pemilihan nilai K
- Membutuhkan waktu komputasi lebih lama
- Rentan terhadap noise pada data dekat

---

## Kesimpulan
WKNN (Weighted K-Nearest Neighbor) merupakan pengembangan dari metode KNN yang bekerja dengan mempertimbangkan kedekatan jarak antar data sebagai bobot dalam proses prediksi. Tidak seperti KNN biasa yang memberikan pengaruh yang sama pada setiap tetangga, WKNN memberikan pengaruh lebih besar pada data yang jaraknya lebih dekat, sehingga hasil prediksi menjadi lebih akurat dan lebih merepresentasikan kondisi sebenarnya.

Metode ini sangat efektif digunakan pada data yang memiliki distribusi tidak merata atau ketika terdapat perbedaan signifikan antar data, karena mampu fokus pada pola lokal di sekitar data yang diprediksi. Selain itu, WKNN bersifat fleksibel karena dapat digunakan baik untuk klasifikasi maupun regresi.

Namun, di balik kelebihannya, WKNN juga memiliki beberapa keterbatasan, seperti kompleksitas perhitungan yang lebih tinggi, sensitivitas terhadap jarak yang sangat kecil, serta ketergantungan pada pemilihan nilai K yang tepat. Oleh karena itu, dalam penggunaannya diperlukan penentuan parameter yang sesuai agar hasil yang diperoleh optimal.


```python
import numpy as np
import pandas as pd

# =========================
# 1. Data Normalisasi
# =========================
data = pd.DataFrame({
    'IPK': [0, 0.5, 1, 0, 0.5, 1],
    'PO':  [0, 0.5, 0, 0, 0.5, 1],
    'JML': [0, 1, 0, 1, 0, 1]
})

# Data yang ingin diprediksi
test = np.array([0, 0.5])  # (IPK, PO)

# =========================
# 2. Hitung Jarak Euclidean
# =========================
def euclidean_distance(row, test):
    return np.sqrt((row[0] - test[0])**2 + (row[1] - test[1])**2)

data['Distance'] = data.apply(lambda row: euclidean_distance(row, test), axis=1)

# =========================
# 3. Urutkan berdasarkan jarak
# =========================
data_sorted = data.sort_values(by='Distance')

print("Data setelah dihitung jaraknya:")
print(data_sorted)

# =========================
# 4. Tentukan K
# =========================
K = 3
neighbors = data_sorted.head(K)

print("\nTetangga terdekat (K=3):")
print(neighbors)

# =========================
# 5. Hitung Bobot (Weight)
# =========================
neighbors['Weight'] = 1 / neighbors['Distance']

print("\nDengan bobot:")
print(neighbors)

# =========================
# 6. Hitung Prediksi WKNN
# =========================
numerator = np.sum(neighbors['Weight'] * neighbors['JML'])
denominator = np.sum(neighbors['Weight'])

prediction = numerator / denominator

print("\nNilai prediksi (sebelum pembulatan):", prediction)

# =========================
# 7. Hasil Akhir
# =========================
final_result = round(prediction)

print("Hasil akhir JML:", final_result)
```
